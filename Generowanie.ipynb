```python
import numpy as np
import pandas as pd
from google.colab import files

np.random.seed(42)

def wygeneruj_realistyczna_noc(harmonogram, start_epoka, podbicie_bpm=0, mnoznik_stresu=1.0):
    dane = []
    epoka_idx = start_epoka

    fazy_z_mikrowybudzeniami = []
    for ilosc, faza in harmonogram:
        for _ in range(ilosc):
            obecna_faza = faza
            if faza in [1, 2, 3] and np.random.rand() < 0.03 * mnoznik_stresu:
                obecna_faza = 0
            fazy_z_mikrowybudzeniami.append(obecna_faza)

    for faza in fazy_z_mikrowybudzeniami:
        if faza == 0:
            bpm = np.random.normal(78 + podbicie_bpm, 6)
            hrv = max(0.1, np.random.normal(2.5 / mnoznik_stresu, 0.8))
            ruch = max(0, np.random.normal(4.5 * mnoznik_stresu, 2.5))
        elif faza == 1:
            bpm = np.random.normal(65 + podbicie_bpm, 3.5)
            hrv = max(0.1, np.random.normal(4.0, 0.8))
            ruch = max(0, np.random.normal(0.8 * mnoznik_stresu, 0.6))
        elif faza == 2:
            bpm = np.random.normal(57 + podbicie_bpm, 1.5)
            hrv = max(0.1, np.random.normal(7.0, 1.2))
            ruch = max(0, np.random.normal(0.05, 0.05))
        elif faza == 3:
            bpm = np.random.normal(70 + podbicie_bpm, 5.0)
            hrv = max(0.1, np.random.normal(1.5, 0.4))
            ruch = max(0, np.random.normal(0.02, 0.02))

        spo2 = np.random.randint(94, 100)
        dane.append([epoka_idx, bpm, spo2, hrv, ruch, faza])
        epoka_idx += 1

    df = pd.DataFrame(dane, columns=['Epoka', 'BPM_Srednie', 'SPO2', 'HRV', 'Energia_Ruchu', 'Faza_Snu'])

    df['BPM_Srednie'] = df['BPM_Srednie'].rolling(window=3, min_periods=1).mean().astype(int)
    df['HRV'] = df['HRV'].rolling(window=2, min_periods=1).mean().round(2)
    df['Energia_Ruchu'] = df['Energia_Ruchu'].round(2)

    return df

dane_uzytkownika = [
    [1, 111, 96, 14.95, 13.32, 0], [2, 98, 98, 5.96, 14.26, 0], [3, 92, 100, 4.54, 13.96, 0],
    [4, 93, 99, 2.92, 14.40, 0], [5, 101, 98, 8.48, 17.72, 0], [6, 118, 98, 9.35, 12.01, 0],
    [7, 119, 99, 6.43, 14.14, 0], [8, 99, 99, 1.95, 12.68, 0]
]
df_user = pd.DataFrame(dane_uzytkownika, columns=['Epoka', 'BPM_Srednie', 'SPO2', 'HRV', 'Energia_Ruchu', 'Faza_Snu'])

harmonogram_noc1 = [
    (15,0), (25,1), (70,2), (20,1), (15,3), (5,0),
    (40,1), (60,2), (20,1), (30,3), (5,0),
    (60,1), (30,2), (40,1), (50,3), (10,0),
    (50,1), (10,2), (40,1), (70,3), (40,1), (20,0)
]

df_noc1_reszta = wygeneruj_realistyczna_noc(harmonogram_noc1, 9)
df_noc1 = pd.concat([df_user, df_noc1_reszta], ignore_index=True)
df_noc1.to_csv("sen_noc1.csv", index=False)
files.download("sen_noc1.csv")

harmonogram_noc2 = [
    (45,0), (50,1), (30,2), (20,1), (10,3), (25,0),
    (60,1), (20,2), (40,1), (30,3), (35,0),
    (80,1), (10,2), (50,3), (20,0), (60,1), (40,0)
]
df_noc2 = wygeneruj_realistyczna_noc(harmonogram_noc2, 1, podbicie_bpm=6, mnoznik_stresu=1.5)
df_noc2.to_csv("sen_noc2.csv", index=False)
files.download("sen_noc2.csv")
```
